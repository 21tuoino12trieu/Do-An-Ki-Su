```python
from openai import OpenAI, AzureOpenAI, AsyncAzureOpenAI
import httpx
import asyncio

# Create an httpx client and pass the SSL certificate path or set to False
httpx_client = httpx.AsyncClient(http2=True, verify=False)

_client = AsyncAzureOpenAI(
    api_version="2024-02-15-preview",
    api_key="REPLACE_WITH_LOCAL_SECRET",
    azure_endpoint="https://10.1.13.168:8443/mbdevai",
    http_client=httpx_client
)

# Define the messages
messages = [
    {"role": "user", "content": "mèo nhà tôi tự nhiên thích ăn thịt chuột, bỏ mẹ rồi"}
]

# Make the API call
response = await _client.chat.completions.create(
    model="azure-chat-engine-gpt-4o-base",
    messages=messages,
    temperature=0,
    top_p=0.001,
    stream=True
)

# Print the response
async for chunk in response:
    if len(chunk.choices) > 0:
        if chunk.choices[0].delta.content:
            print(chunk.choices[0].delta.content, end="", flush=True)
            await asyncio.sleep(0.01)

# Close the client when done
await httpx_client.aclose()
```
