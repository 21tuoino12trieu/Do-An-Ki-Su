```python
import requests
import json

def chat(question: str):
    # API configuration
    url = "http://beegen-credit-bo-api.tanzu-dev.mbbank.com.vn/v1/chat/completions"
    headers = {
        'Content-Type': 'application/json'
    }
    
    payload = {
        "bot_id": "31793bac-be65-4538-bc97-a42e87a18ede",
        "messages": [
            {
                "role": "user",
                "content": question
            }
        ],
        "role": "L1",
        "topic": "default",
    }
    
    response = requests.post(url, headers=headers, json=payload)
    
    print(response.json())
```

```python
chat("xin chào")
```
